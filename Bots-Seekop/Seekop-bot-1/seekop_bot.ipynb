import re

files = ["main.py", "seekop_api.py", "sheets.py", "config.py"]

for fname in files:
    with open(fname, "r", encoding="utf-8") as f:
        contenido = f.read()
    # Elimina cualquier caracter fuera del rango ASCII/Latin
    contenido = re.sub(r'[^\x00-\x7F\xc0-\xff]', '', contenido)
    with open(fname, "w", encoding="utf-8") as f:
        f.write(contenido)

print("Listo, todos los archivos corregidos.")
